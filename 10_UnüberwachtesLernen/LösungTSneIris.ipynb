```python
import pandas as pd
df = pd.read_csv("../Datensätze/iris.csv")
X = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].values
y = df[["type"]].values
```

## Dimensionsreduktion mit t-SNE

```python
from sklearn.manifold import TSNE
tsne = TSNE(n_components=2, random_state=0, perplexity = 60)
#tsne = TSNE(random_state=0)
X_2d = tsne.fit_transform(X)
```

```python
%matplotlib inline
from matplotlib import pyplot as plt

setosa = X_2d[df["type"]==1]
versicolor = X_2d[df["type"]==2]
virginica = X_2d[df["type"]==3]

plt.scatter(setosa[:,0], setosa[:,1], c='r', label = "setosa")
plt.scatter(versicolor[:,0], versicolor[:,1], c='g', label = "versicolor")
plt.scatter(virginica[:,0], virginica[:,1], c='b', label = "virginica")

plt.legend()
plt.show()
```
