### MNIST-Daten über Keras laden

```python
from tensorflow.keras.datasets import mnist

# Datensatz laden
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Pixelwerte auf den Bereich [0, 1] skalieren
X_train = X_train / 255.0
X_test = X_test / 255.0

X = X_train.reshape(-1, 784)[:5000,:]
Y = y_train[:5000]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_transformed = scaler.fit_transform(X)
```

### Dimensionalitäsreduktion mit MDS

```python
from sklearn.manifold import MDS
mds = MDS(n_components=2, random_state=0, n_jobs = -1)
X_2d = mds.fit_transform(X)
```

Projizierte Datensätze plotten

```python
%matplotlib inline
from matplotlib import pyplot as plt

plt.scatter(X_2d[:,0], X_2d[:,1], c=Y, s=15)
plt.savefig("scatter_mds_mnist.svg", format="svg", bbox_inches="tight")
plt.show()
```
