```python
import pandas as pd
df = pd.read_csv("../Datensätze/fahrzeug_preis.csv")
X = df[["yearOfRegistration", "price"]]
df.head()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.scatter(df["yearOfRegistration"], df["price"])
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_transformed = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters = 3)
model.fit(X_transformed)
# Label vom Clusterer abfragen
labels = model.labels_

#Centroide vom Clusterer abfragen
centers = model.cluster_centers_
# Die Center müssen wir aus dem skalierten Raum zurücktransformieren 
centers_transformed = scaler.inverse_transform(centers)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

# Punkte ausgeben
plt.scatter(df["yearOfRegistration"], df["price"], c = labels, alpha = 0.5, s = 30)

# Centroide ausgeben
plt.scatter(
    centers_transformed[:, 0], 
    centers_transformed[:, 1], 
    c = range(len(centers_transformed)), 
    marker = "X", edgecolor = 'black',
    s = 200)
plt.show()
```
