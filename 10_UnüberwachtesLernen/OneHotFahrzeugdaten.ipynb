```python
import pandas as pd
df = pd.read_csv("../Datensätze/fahrzeug_preis.csv")
df.head()
```

## Daten  aufbereiten
* One Hot Encoding durchführen
* Unnötige Spalten verwerfen

```python
df = pd.get_dummies(df, columns = ["fuelType"])
df =df.drop(labels=["fuelType_lpg"],axis = 1)
df =df.drop(labels=["model", "name"],axis = 1)

X = df[["price", "yearOfRegistration", "powerPS", "kilometer", 
        "fuelType_benzin", "fuelType_diesel", "fuelType_hybrid"]].values

df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_transformed = scaler.fit_transform(X)
```

## Multi-Dimensional Scaling

```python
from sklearn.manifold import MDS
mds = MDS(n_components=2, random_state=0)
X_2d = mds.fit_transform(X_transformed)
```

```python
%matplotlib inline
from matplotlib import pyplot as plt
plt.scatter(X_2d[:,0], X_2d[:,1])

plt.show()
```
