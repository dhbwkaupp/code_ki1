```python
import pandas as pd
df = pd.read_csv("../Datensätze/fahrzeug_preis.csv")
df.head()
```

## Daten aufbereiten
* One Hot Encoding durchführen
* Unnötige Spalten verwerfen

```python
df = pd.get_dummies(df, columns = ["fuelType"])
df =df.drop(labels=["fuelType_lpg"],axis = 1)
df =df.drop(labels=["model", "name"],axis = 1)

X = df[["price", "yearOfRegistration", "powerPS", "kilometer", 
        "fuelType_benzin", "fuelType_diesel", "fuelType_hybrid"]].values

df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_transformed = scaler.fit_transform(X)
```

## Multi-Dimensional Scaling

```python
from sklearn.manifold import MDS
mds = MDS(n_components=2, random_state=0)
X_2d = mds.fit_transform(X_transformed)
```

```python
%matplotlib inline
from matplotlib import pyplot as plt
plt.scatter(X_2d[:,0], X_2d[:,1])
plt.show()
```

## Elbow Methode

```python
from sklearn.cluster import KMeans
scores = []
for n in range(2, 10):
    model = KMeans(n_clusters = n)
    model.fit(X_transformed)
    scores.append(model.inertia_)
scores
```

```python
%matplotlib inline

import matplotlib.pyplot as plt

plt.plot(range(2, 10), scores)
plt.show()
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters = 5)
model.fit(X_transformed)

labels = model.labels_
centers = model.cluster_centers_
```

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np

colorlist = ['r', 'g', 'b', 'y', 'm']

for i in np.unique(labels):
    plt.scatter(X_2d[labels==i,0], X_2d[labels==i,1], 
                c=colorlist[i], label=i)
plt.legend()
plt.show()
```
