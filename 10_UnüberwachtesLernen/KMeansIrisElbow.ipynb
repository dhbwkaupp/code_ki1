```python
import pandas as pd
df = pd.read_csv("../Datensätze/iris.csv")
X = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].values
y = df[["type"]].values
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_transformed = scaler.fit_transform(X)
```

```python
from sklearn.manifold import MDS
mds = MDS(n_components=2, random_state=0)
X_2d = mds.fit_transform(X)
```

```python
%matplotlib inline
from matplotlib import pyplot as plt

plt.scatter(X_2d[:,0], X_2d[:,1], c=df["type"])
plt.show()
```

## Quadratische Abstände ermitteln für verschiedene k

```python
from sklearn.cluster import KMeans
scores = []
for n in range(2, 10):
    model = KMeans(n_clusters = n)
    model.fit(X_transformed)
    scores.append(model.inertia_)
```

## Elbow-Graph zeichnen

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.plot(range(2, 10), scores)
plt.show()
```

## k-Means mit dem abgelesenen k

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters = 3)
model.fit(X_transformed)

labels = model.labels_
```

```python
%matplotlib inline

import matplotlib.pyplot as plt

plt.scatter(X_2d[:,0], X_2d[:,1], c = labels, alpha = 0.7)
plt.show()
```
