```python
import pandas as pd

df = pd.read_csv("../Datensätze/qualitaet_spritzguss.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split

# Welche Spalten sollen zur Vorhersage verwendet werden
X = df[["Druck", "Temperatur"]].to_numpy()
y = df["Erfolg"].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 1, test_size = 0.25)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion='entropy')
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```

```python
# Hinweis: Damit dieser Befehl funktioniert, muss die 
# "helper.py" - Datei im selben Ordner liegen, wie das
# aktuelle Jupyter Notebook

from helper import plot_classifier

# Trainings-Daten plotten
plot_classifier(model, X_train, y_train, proba = False, 
                xlabel = "Druck", ylabel = "Temperatur", 
                cmapName = "bwr")
```

```python
# Testdaten plotten

plot_classifier(model, X_test, y_test, proba = False, xlabel = "Druck", ylabel = "Temperatur", cmapName = "bwr")
```

### Vorhersage machen
Mit dem einmal trainierten Modell können wir jetzt Vorhersagen für bisher unbekannte

```python
import numpy as np
X_wert = np.array([[830, 240]])
ergebnis = model.predict(X_wert)
print( ergebnis[0] )
```
