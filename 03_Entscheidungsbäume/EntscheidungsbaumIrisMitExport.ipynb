```python
import pandas as pd

df = pd.read_csv("../Datensätze//iris.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split

# Welche Spalten sollen zur Vorhersage verwendet werden
X = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].to_numpy()
y = df["type"].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 1, test_size = 0.25)
```

## Entscheidungsbaum mit begrenzter Tiefe

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion='entropy', max_depth = 4)
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(model, 
        feature_names = ["sepal length", "sepal width", "petal length", "petal width"],
        class_names = ["Setosa ", "Versicolour", "Virginica"],
        filled = True,
        rounded = True
    )
plt.show()
```

## Entscheidungsbaum mit begrenzter Tief und minimaler Datensatzzahl

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion='entropy', max_depth = 4, min_samples_leaf = 3)
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(model, 
        feature_names = ["sepal length", "sepal width", "petal length", "petal width"],
        class_names = ["Setosa ", "Versicolour", "Virginica"],
        filled = True,
        rounded = True
    )
plt.show()
```

## Klassifikator speichern
Für einfache Zwecke und vertrauenswürdige Quellen ist das eingebaute `pickle` prima.

```python
import pickle
file_name = "modell.pickle"
pickle.dump(model, open(file_name, "wb"))
```

## Klassifikator wieder laden und neuen Datensatz klassifizieren

```python
import numpy as np
loaded_model = pickle.load(open(file_name, "rb"))
messwerte = [[5.2, 3.4, 1.5, 0.2]]
messwerte = np.array(messwerte)
loaded_model.predict(messwerte)
```
