```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Ebene definieren: ax + by + cz = d
a, b, c = 2, -3, 1  # Normalenvektor (2, -3, 1)
d = 6               # Abstand der Ebene

# Funktion, um z-Werte für die Ebene zu berechnen
def plane_z(x, y, a, b, c, d):
    return (d - a * x - b * y) / c

# Punkte in der Ebene erzeugen
x = np.linspace(-5, 5, 10)  # x-Koordinaten
y = np.linspace(-5, 5, 10)  # y-Koordinaten
x, y = np.meshgrid(x, y)    # Gitter erzeugen
z = plane_z(x, y, a, b, c, d)

# Zufällige Punkte generieren
num_points = 50
points = np.random.uniform(-5, 5, (num_points, 3))  # Zufällige Punkte im Bereich -5 bis 5

# Plot erstellen
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Ebene plotten
ax.plot_surface(x, y, z, alpha=0.5, color='blue', edgecolor='k')

# Punkte plotten
ax.scatter(points[:, 0], points[:, 1], points[:, 2], color='red', label='Punkte')

# Normalenvektor visualisieren
normal_vector = np.array([a, b, c])
origin = np.array([0, 0, d / c])  # Punkt auf der Ebene
ax.quiver(origin[0], origin[1], origin[2], normal_vector[0], normal_vector[1], normal_vector[2],
          color='green', label='Normalenvektor', length=4)

# Labels und Legende
ax.set_xlabel('X-Achse')
ax.set_ylabel('Y-Achse')
ax.set_zlabel('Z-Achse')
ax.legend()

# Plot anzeigen
plt.savefig("svm_normalenvektor.svg", format="svg", bbox_inches="tight")
plt.show()
```
