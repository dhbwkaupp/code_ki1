```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the 3D function (e.g., a parabolic bowl)
def func(x, y):
    return x**2 + 2*y**2

# Generate data for the 3D plot
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = func(X, Y)

# Simulate gradient descent steps
steps = 10
learning_rate = 0.3
x_gd, y_gd = [-4], [4]  # Starting point
z_gd = [func(x_gd[0], y_gd[0])]

for _ in range(steps):
    grad_x = 2 * x_gd[-1]  # Derivative w.r.t. x
    grad_y = 2 * y_gd[-1]  # Derivative w.r.t. y
    x_new = x_gd[-1] - learning_rate * grad_x
    y_new = y_gd[-1] - learning_rate * grad_y
    z_new = func(x_new, y_new)
    x_gd.append(x_new)
    y_gd.append(y_new)
    z_gd.append(z_new)

# Plot the 3D surface
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

# Plot the gradient descent path
ax.plot(x_gd, y_gd, z_gd, color='red', marker='o', label='Pfad des Gradientenabstiegs')

# Label the axes
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Add legend
ax.legend()
plt.show()
```

```python
# Define a multidimensional function (e.g., a 2D paraboloid)
def func(x, y):
    return x**2 + 2*y**2

# Define the gradient of the function
def gradient(x, y):
    grad_x = 2 * x
    grad_y = 4 * y
    return np.array([grad_x, grad_y])

# Generate data for the function
x = np.linspace(-4, 4, 20)
y = np.linspace(-4, 4, 20)
X, Y = np.meshgrid(x, y)
Z = func(X, Y)

# Compute gradient vectors
U, V = gradient(X, Y)

# Plot the function as a contour plot
plt.figure(figsize=(8, 6))
contour = plt.contour(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(contour, label="Funktionswert")

# Add gradient vectors using quiver
plt.quiver(X, Y, U, V, color="red", scale=20, label="Gradientenvektoren")

# Label the axes
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)

# Add a legend
plt.legend()

# Show the plot
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
# Python-Implementierung einer einfachen Backpropagation für ein neuronales Netz mit einer versteckten Schicht

import numpy as np

# Sigmoid-Aktivierungsfunktion und ihre Ableitung
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Trainingsdaten (X: Eingaben, y: Zielwerte)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])  # XOR-Eingaben
y = np.array([[0], [1], [1], [0]])  # XOR-Ausgaben

# Initialisierung der Gewichte
np.random.seed(42)
input_layer_neurons = X.shape[1]  # Anzahl der Eingabeneuronen
hidden_layer_neurons = 2          # Anzahl der Neuronen in der versteckten Schicht
output_neurons = 1                # Anzahl der Ausgabeneuronen

# Zufällige Gewichtsinitalisierung
weights_input_hidden = np.random.uniform(size=(input_layer_neurons, hidden_layer_neurons))
weights_hidden_output = np.random.uniform(size=(hidden_layer_neurons, output_neurons))

# Lernrate
learning_rate = 0.5

# Training
epochs = 10000
for epoch in range(epochs):
    # Vorwärtsdurchlauf
    hidden_layer_input = np.dot(X, weights_input_hidden)
    hidden_layer_output = sigmoid(hidden_layer_input)

    output_layer_input = np.dot(hidden_layer_output, weights_hidden_output)
    predicted_output = sigmoid(output_layer_input)

    # Fehlerberechnung
    error = y - predicted_output

    # Rückwärtsdurchlauf (Backpropagation)
    d_predicted_output = error * sigmoid_derivative(predicted_output)
    error_hidden_layer = d_predicted_output.dot(weights_hidden_output.T)
    d_hidden_layer = error_hidden_layer * sigmoid_derivative(hidden_layer_output)

    # Gewichts-Updates
    weights_hidden_output += hidden_layer_output.T.dot(d_predicted_output) * learning_rate
    weights_input_hidden += X.T.dot(d_hidden_layer) * learning_rate


    # Optional: Fehlerausgabe pro Epoche
    if epoch % 1000 == 0:
        loss = np.mean(np.abs(error))
        print(f"Epoche {epoch}, Fehler: {loss:.4f}")

# Vorhersage nach dem Training
print("\nTrainierte Ausgaben:")
print(predicted_output)
```
