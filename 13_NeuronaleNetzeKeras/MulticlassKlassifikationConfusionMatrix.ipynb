# Laden der Fashion MNIST Daten

```python
import numpy as np
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.datasets import fashion_mnist
 
(X_train, y_train), (X_test, y_test_original) = fashion_mnist.load_data()
 
# Aufbereiten der eingelesenen Daten
X_train = X_train / 255.0
X_test = X_test / 255.0

X_train = X_train.reshape(60000, 784) # Umwandeln von 28x28 in "flache" 784 Pixel
X_test = X_test.reshape(10000, 784) # Umwandeln von 28x28 in "flache" 784 Pixel

# One Hot encoding
y_train = to_categorical(y_train)
y_test = to_categorical(y_test_original)
```

```python
# Ausgabe eines One Hot encodierten Datensatzes
print(y_train[0])
```

# Neuronales Netz erzeugen

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

model = Sequential()
model.add(Input(shape=(784,)))
# Hidden
model.add(Dense(512, activation="relu"))
model.add(Dense(256, activation="relu"))
model.add(Dense(128, activation="relu"))
# Output
model.add(Dense(10, activation="softmax"))

# Für Multiclass-Probleme nimmt man als loss-Funktion oft categorical_crossentropy  
model.compile(optimizer="sgd", loss="categorical_crossentropy", metrics=["accuracy"])
```

# Neuronales Netz trainieren

```python
model.fit(X_train, y_train, epochs = 100, batch_size = 600, verbose = 1, validation_split=0.1)
```

# Genauigkeit auf Testdaten

```python
model.evaluate(X_test, y_test)
```

# Confusion Matrix erstellen

```python
from sklearn import metrics
import numpy as np

# Wir berechnen alle Vorhersagen mit einem Methodenaufruf
predictions = model.predict(X_test)

# vorhersagen ist ein 10000x10 Feld. Mit axis können wir
# wählen auf welcher Achse das argmax angewendet wird
predictions = np.argmax(predictions, axis = 1)

confusion_matrix = metrics.confusion_matrix(y_test_original, predictions)
```

```python
import matplotlib.pyplot as plt
import numpy
from sklearn import metrics

cm_display = metrics.ConfusionMatrixDisplay(
    confusion_matrix = confusion_matrix, 
    display_labels = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat", 
                      "Sandal", "Shirt","Sneaker", "Bag", "Ankle boot"],
)
cm_display.plot(xticks_rotation='vertical', colorbar=False) # Labels an x-Achse senkrecht
plt.show()
```
