# Laden der Fashion MNIST Daten
Wir machen das praktisch gleich wie bei den originalen MNIST-Daten.

```python
import numpy as np
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.datasets import fashion_mnist

 
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
 
# Aufbereiten der eingelesenen Daten
X_train = X_train / 255.0
X_test = X_test / 255.0

X_train = X_train.reshape(60000, 784) # Umwandeln von 28x28 in "flache" 784 Pixel
X_test = X_test.reshape(10000, 784) # Umwandeln von 28x28 in "flache" 784 Pixel

# Wir wenden den boolschen Vergleich auf alle Datensätze an.
# Jetzt steht an Label eine 1 (true), wenn es ein T-Shirt ist, und sonst eine 0 (false).
y_train = (y_train == 0)
y_test = (y_test == 0)
```

# Neuronales Netz erzeugen

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# Bei Bedarf: Seed setzen, bevor das Modell erzeugt wird
tf.keras.utils.set_random_seed(17)

# Mit Sequential erzeugt man ein Modell, dem man dann einzelne Layer hinzufügen kann
model = Sequential()

# Inputlayer mit Shape der Eingabedaten hinzufügen
model.add(Input(shape=(784,)))

# Dense Layer sind "fully Connected". Alle Neuronen sind mir allen Inputs und Outputs verbunden.
model.add(Dense(100, activation="relu"))
model.add(Dense(1, activation="sigmoid"))

# sgd = stochatic gradient descent
model.compile(optimizer="sgd", loss="binary_crossentropy")
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(
    model,
    #to_file="model.svg",
    show_shapes=True,
    show_layer_names=True
)
```

# Neuronales Netz trainieren

```python
#model.fit(X_train, y_train, epochs = 10, batch_size = 600, verbose = 1)

history = model.fit(
    X_train, y_train,
    epochs=300,
    batch_size=600,
    validation_split=0.25,
    verbose=1
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

lossMonitor = np.array(history.history['loss'])
valLossMonitor = np.array(history.history['val_loss'])
counts = np.arange(lossMonitor.shape[0])

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot(counts, lossMonitor, 'k', label='Trainingsdaten')
ax.plot(counts, valLossMonitor, 'r:', label='Validierungsdaten')
ax.set_xlabel('Lernzyklus')
ax.set_ylabel('Fehler')
ax.set_xticks(counts[::50])
#ax.set_xticks(counts)
ax.set_ylim(0, 1.0)
ax.legend()
plt.show()
```
