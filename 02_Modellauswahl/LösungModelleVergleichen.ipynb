# Analyse des Benzinverbrauchs von Fahrzeugen

## Daten einlesen

```python
import pandas as pd
df = pd.read_csv("../Datensätze/auto-mpg.csv")
df.head()
```

## Aufteilung der Datensätze

```python
from sklearn.model_selection import train_test_split
X = df[["weight", "horsepower"]].to_numpy()
Y = df[["mpg"]].to_numpy()
X_train, X_test, Y_train, Y_test = train_test_split(X,Y)
```

## Training

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,Y_train)
```

## Vorhersage für die Testdaten berechnen

```python
y_test_predicted = model.predict(X_test)
```

## Bestimmtheitsmaß berechnen mit Funktion r2_score

```python
from sklearn.metrics import r2_score

r2 = r2_score(Y_test, y_test_predicted)
print(r2)
```

## Kurzvariante: Bestimmtheitsmaß über Methodenaufruf

```python
print(model.score(X_test, Y_test))
```
