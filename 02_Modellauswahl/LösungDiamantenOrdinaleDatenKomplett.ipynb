# Analyse Diamantendaten mit Datenaufbereitung

```python
import pandas as pd
df = pd.read_csv("../Datensätze/diamonds.csv")
df.head()
```

```python
print(df["cut"].value_counts())
print(df["cut"].value_counts().count())
```

```python
pd.set_option('future.no_silent_downcasting', True)

replace_map = {'cut': {'Fair': 1,  'Good':2, 'Very Good':3, 'Premium':4, 'Ideal':5}}
df = df.replace(replace_map)
df['cut'] = df['cut'].astype(int)

replace_map = {'color': {'D': 7,  'E':6, 'F':5, 'G':4, 'H':3, 'I':2, 'J':1}}
df = df.replace(replace_map)
df['cut'] = df['color'].astype(int)

replace_map = {'clarity': {'FL':11,  'IF':10, 'VVS1':9, 'VVS2':8, 'VS1':7, 
                           'VS2':6, 'SI1':5, 'SI2':4, 'I1':3, 'I2':2, 'I3':1}}
df = df.replace(replace_map)
df['cut'] = df['clarity'].astype(int)

df.head()
```

```python
X = df[["carat", "cut", "color", "clarity"]].values
Y = df[["price"]].values
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
import numpy as np

scores = cross_val_score(LinearRegression(), X, Y, 
                         cv = KFold(n_splits = 10, shuffle = True))

print("Ergebnis Karat: " + str(np.mean(scores)))
```
