# k-fache Kreuzvalidierung mit der Hilfsfunktion cross_val_score


```python
import pandas as pd
df = pd.read_csv("../Datensätze/auto-mpg.csv")
df.head()
```

```python
X = df[["weight", "horsepower"]].values
Y = df[["mpg"]].values
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
import numpy as np

#scores = cross_val_score(LinearRegression(), X, Y) # Default-KFold (3-fach)

kf = KFold(n_splits = 10, shuffle = True)
scores = cross_val_score(LinearRegression(), X, Y, cv = kf)
#scores = cross_val_score(LinearRegression(), X, Y, cv = KFold(n_splits = 10, shuffle = True))

print(scores)
print(np.mean(scores))
```
