# Analyse Diamantendaten

```python
import pandas as pd
df = pd.read_csv("../Datensätze/diamonds.csv")
df.head()
```

```python
X = df[["carat"]].values
Y = df[["price"]].values
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
import numpy as np

scores = cross_val_score(LinearRegression(), X, Y, 
                         cv = KFold(n_splits = 10, shuffle = True))

print("Ergebnis Karat: " + str(np.mean(scores)))
```

```python
X = df[["x", "y", "z"]].values
Y = df[["price"]].values
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
import numpy as np

scores = cross_val_score(LinearRegression(), X, Y, 
                         cv = KFold(n_splits = 10, shuffle = True))

print("Ergebnis Abmessungen: " + str(np.mean(scores)))
```
