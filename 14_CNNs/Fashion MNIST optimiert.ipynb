# Laden der Fashion MNIST Daten


```python
import gzip
import numpy as np
from tensorflow.keras.utils import to_categorical

# Funktion zum Einlesen aller Bilder aus einer Datei
def open_images(filepath):
    with gzip.open(filepath, "rb") as file:
        data = file.read()
        return np.frombuffer(data, dtype=np.uint8, offset=16)\
            .reshape(-1, 28, 28)\
            .astype(np.float32)

# Funktion zum Einlesen aller Label aus einer Datei
def open_labels(filepath):
    with gzip.open(filepath, "rb") as file:
        data = file.read()
        return np.frombuffer(data, dtype=np.uint8, offset=8)

# Einlesen
X_train = open_images(base_path + "/Datensätze/fashion/train-images-idx3-ubyte.gz")
y_train = open_labels(base_path + "/Datensätze/fashion/train-labels-idx1-ubyte.gz")
X_test = open_images(base_path + "/Datensätze/fashion/t10k-images-idx3-ubyte.gz")
y_test = open_labels(base_path + "/Datensätze/fashion/t10k-labels-idx1-ubyte.gz")

# One Hot encoding
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

```python
print(X_train.shape)
```

# CNN erzeugen

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, MaxPooling2D, Conv2D, Flatten, Dropout

model = Sequential()
model.add(Input(shape=(28, 28, 1)))

model.add(Conv2D(32, kernel_size=(3, 3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))
#model.add(Dropout(0.25))
model.add(Flatten())
model.add(Dense(100, activation="sigmoid"))

# Man kann mehrere Dropout-Layer haben.
#model.add(Dropout(0.25))

# Für 10 Klassen braucht man 10 Ausgabe-Neuronen
model.add(Dense(10, activation="softmax"))

model.compile(optimizer="sgd", loss="categorical_crossentropy", metrics=["accuracy"])
```

# Neuronales Netz trainieren

```python
model.fit(X_train, y_train, epochs = 100, batch_size = 32, verbose = 1)
```

# Genauigkeit auf Testdaten

```python
model.evaluate(X_test, y_test)
```
