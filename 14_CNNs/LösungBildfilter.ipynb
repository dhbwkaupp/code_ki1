```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Bild laden
bild = cv2.imread("campus.png")
bild = cv2.cvtColor(bild, cv2.COLOR_BGR2RGB)
```

```python
import numpy as np

def sharpen_filter():
    """Schärfungsfilter"""
    return np.array([
        [ 0, -1,  0],
        [-1,  5, -1],
        [ 0, -1,  0]
    ])

def gaussian_filter():
    """3x3 Gauß-Filter zur Glättung"""
    return np.array([
        [1, 2, 1],
        [2, 4, 2],
        [1, 2, 1]
    ]) / 16

def mean_filter():
    """3x3 Mittelwertfilter zur Glättung"""
    return np.array([
        [1, 1, 1],
        [1, 1, 1],
        [1, 1, 1]
    ]) / 9

def sobel_x_filter():
    """Kantenfilter X (vertikale Kanten)"""
    return np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ])

def sobel_y_filter():
    """Kantenfilter Y (horizontale Kanten)"""
    return np.array([
        [-1, -2, -1],
        [ 0,  0,  0],
        [ 1,  2,  1]
    ])

def laplace_filter():
    """Laplace-Kantenfilter"""
    return np.array([
        [ 0, -1,  0],
        [-1,  4, -1],
        [ 0, -1,  0]
    ])

def emboss_filter():
    """Relief-/Emboss-Effekt"""
    return np.array([
        [-2, -1, 0],
        [-1,  1, 1],
        [ 0,  1, 2]
    ])
```

```python
# Am besten erst Gaussian
kernel = gaussian_filter()
gefiltert = cv2.filter2D(bild, -1, kernel)
kernel = emboss_filter()
gefiltert = cv2.filter2D(bild, -1, kernel)

# Ergebnis speichern
cv2.imwrite("filter_campus_.png", cv2.cvtColor(gefiltert, cv2.COLOR_RGB2BGR))

# Original- und Ergebnisbild anzeigen
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.imshow(bild)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(gefiltert)
plt.title("Gefiltert")
plt.axis("off")

plt.show()
```
