```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Zielfunktion
def f(x, y):
    return (x - 2)**2 + (y + 1)**2 + np.sin(3*x) * np.cos(2*y)
```

```python
# 3D-Darstellung der Funktion
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_title('Zielfunktion')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x,y)')

plt.show()
```

```python
np.random.seed(42)

n_particles = 30
n_iter = 50

w = 0.7
c1 = 1.5
c2 = 1.5

# Initialisierung
pos = np.random.uniform(-5, 5, (n_particles, 2))
vel = np.random.uniform(-1, 1, (n_particles, 2))

# Persönliche und globale Bestwerte
pbest = pos.copy()
pbest_val = np.array([f(x, y) for x, y in pos])

gbest_idx = np.argmin(pbest_val)
gbest = pbest[gbest_idx].copy()
gbest_val = pbest_val[gbest_idx]

history = []

for _ in range(n_iter):

    r1 = np.random.rand(n_particles, 2)
    r2 = np.random.rand(n_particles, 2)

    # Geschwindigkeitsupdate
    vel = (
        w * vel
        + c1 * r1 * (pbest - pos)
        + c2 * r2 * (gbest - pos)
    )

    # Positionsupdate
    pos += vel

    # Partikel im Suchraum halten
    pos[:, 0] = np.clip(pos[:, 0], -5, 5)
    pos[:, 1] = np.clip(pos[:, 1], -5, 5)

    # Bewertung der aktuellen Positionen
    values = np.array([f(x, y) for x, y in pos])

    # Persönliche Bestwerte aktualisieren
    improved = values < pbest_val
    pbest[improved] = pos[improved]
    pbest_val[improved] = values[improved]

    # Globalen Bestwert aktualisieren
    best_idx = np.argmin(pbest_val)

    if pbest_val[best_idx] < gbest_val:
        gbest = pbest[best_idx].copy()
        gbest_val = pbest_val[best_idx]

    history.append(pos.copy())

# Ausgabe
x_min, y_min = gbest

print("Globales Minimum gefunden bei:")
print(f"x = {x_min:.4f}")
print(f"y = {y_min:.4f}")
print(f"f(x,y) = {gbest_val:.4f}")
```

```python
# Visualisierung der Partikel

fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6)

final_pos = history[-1]

ax.scatter(
    final_pos[:,0],
    final_pos[:,1],
    f(final_pos[:,0], final_pos[:,1]),
    color='red',
    s=40,
    label='Partikel'
)

ax.scatter(
    gbest[0],
    gbest[1],
    gbest_val,
    color='black',
    marker='*',
    s=200,
    label='Globales Minimum'
)

ax.legend()
ax.set_title('Partikelschwarmoptimierung')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x,y)')

plt.show()
```
