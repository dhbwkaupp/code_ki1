# Bild laden und für VGG16 vorverarbeiten

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.vgg16 import preprocess_input, VGG16, decode_predictions
image = load_img("tier.jpg", target_size=(224, 224))
image = img_to_array(image)
image = image.reshape((1, 224, 224, 3))
image = preprocess_input(image)
```

# VGG16-Modell laden (Teil der KERAS-API)

```python
model = VGG16()
pred = model.predict(image)
```

```python
decode_predictions(pred)
```
