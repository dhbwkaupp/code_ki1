# Bild laden und ins richtige Format bringen


```python
from PIL import Image

#image = Image.open("tier.jpg")
image = Image.open("cat-551554_640.jpg")
```

# Aufbereitung für VGG16

```python
import numpy as np
from tensorflow.keras.applications.vgg16 import preprocess_input

image = image.resize((224, 224), Image.LANCZOS)
image = np.asarray(image).reshape(1, 224, 224, 3)
image = preprocess_input(image)
```

# Daten durch VGG16 schicken

```python
from tensorflow.keras.applications.vgg16 import VGG16

vgg16 = VGG16(include_top=False, input_shape=(224, 224, 3))
X_nach_vgg = vgg16.predict(image)
```

# Unser Netz laden und Zwischenergebnis klassifizieren

```python
from keras.models import load_model

model = load_model("ErkennungHundKatze.keras")
model.predict(X_nach_vgg)
```
