## Daten einlesen
Für die bessere Vergleichbarkeit von Ergebnissen ist der MNIST-Datensatz bereits in Trainings- und Testdaten aufgeteilt.

Der Datensatz ist **in der ML-Bibliothek Keras enthalten**. Über die Funktion `mnist.load_data()` aus Keras werden die Daten automatisch geladen. **Beim ersten Aufruf** wird der Datensatz aus dem Internet heruntergeladen und **lokal auf dem Rechner gespeichert**. Bei späteren Aufrufen werden die bereits gespeicherten Dateien verwendet, sodass kein erneuter Download erforderlich ist.

Der Datensatz enthält:
* 60.000 Trainingsbilder mit zugehörigen Labels
* 10.000 Testbilder mit zugehörigen Labels
* Graustufenbilder mit einer Größe von 28 × 28 Pixeln
* 10 Klassen (Ziffern von 0 bis 9)

Nach dem Einlesen werden die Pixelwerte durch Division durch 255 auf den Bereich $[0,1]$ normiert.

```python
from tensorflow.keras.datasets import mnist

# Datensatz laden
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Pixelwerte auf den Bereich [0, 1] skalieren
X_train = X_train / 255.0
X_test = X_test / 255.0
```

Wir schauen uns einen einzelnen Datensatz an:

```python
%matplotlib inline

import matplotlib.pyplot as plt

print("Label: %d" % y_train[0])
plt.imshow(X_train[0])
plt.show()
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()

# Klassifikatoren können nichts mit 28x28-Matrizen anfangen. 
# Wir formatieren die Daten um, dass jede Eingabe zu einem 
# eindimensionalen Vektor der Länge 784 wird. (28*28 = 784)
model.fit(X_train.reshape(-1, 784), y_train)
```

Einige Klassifikatoren brauchen ziemlich lange für die Klassifikation. Deswegen kann es Sinn machen, für die Vorauswahl die _Menge der Testdaten zu reduzieren_ (z.B. auf 1000 Datensätze)

```python
#Score berechnen mit allen 10.000 Datensätzen
model.score(X_test.reshape(-1, 784), y_test)

#Score berechnen mit nur 1000 Datensätzen
model.score(X_test.reshape(-1, 784)[:1000,:], y_test[:1000])
```
