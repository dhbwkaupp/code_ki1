```python
import pandas as pd

df = pd.read_csv("../Datensätze/qualitaet_spritzguss.csv")
df.head()
```

```python
# Welche Spalten sollen zur Vorhersage verwendet werden
X = df[["Druck", "Temperatur"]].values
y = df["Erfolg"].values
```

Wir können eine Bearbeitungs-Pipeline definieren. Hier sagen wir z.B., dass auf den Daten erst der StandardScaler ausgeführt werden soll, dann der RandomForestClassifier.

Jede Komponente der Pipeline bekommt einen Namen.

```python
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("rf", RandomForestClassifier())
])
```

**GridSearchCV** probiert verschiedene Parameter-Konfigurationen für Lernmodelle aus und gibt am Ende die Parameterkombination mit dem besten Score aus. Die Scores werden dabei schon automatisch mit einer k-fachen Kreuzvalidierung bestimmt.

```python
from sklearn.model_selection import GridSearchCV

clf = GridSearchCV(pipeline, param_grid = {
    # hier muss der Name stehen, den wir dem Objekt in der Pipeline
    # zugeordnet haben (siehe oben), dann zwei Unterschritte, dann
    # der Name des Parameters. Anschließend kommt die Liste mit den Werten,
    # die ausprobiert werden sollen.
    "rf__n_estimators": range(470,500,1)
})
clf.fit(X,y)

print(clf.best_params_)
```
