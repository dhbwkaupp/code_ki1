```python
import pandas as pd

df = pd.read_csv("../Datensätze/iris.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split

# Welche Spalten sollen zur Vorhersage verwendet werden
X = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].to_numpy()
y = df["type"].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 1, test_size = 0.25)
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(criterion = "entropy", n_estimators = 30)
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```

## Adaboost

## Entscheidungsbaum mit begrenzter Tief und minimaler Datensatzzahl

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
#AdaBoostClassifier?
#model = AdaBoostClassifier()
model = AdaBoostClassifier(estimator = DecisionTreeClassifier(max_depth=3))
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```
