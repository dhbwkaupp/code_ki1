```python
import pandas as pd

df = pd.read_csv("../Datensätze/iris.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split

# Welche Spalten sollen zur Vorhersage verwendet werden
X = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].to_numpy()
y = df["type"].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 1, test_size = 0.25)
```

## Entscheidungsbaum mit begrenzter Tiefe

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion='entropy', max_depth = 4)
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(model, 
        feature_names = ["sepal length", "sepal width", "petal length", "petal width"],
        class_names = ["Setosa ", "Versicolour", "Virginica"],
        filled = True,
        rounded = True
    )
plt.show()
```

## Entscheidungsbaum mit begrenzter Tief und minimaler Datensatzzahl

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion='entropy', max_depth = 4, min_samples_leaf = 3)
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(model, 
        feature_names = ["sepal length", "sepal width", "petal length", "petal width"],
        class_names = ["Setosa ", "Versicolour", "Virginica"],
        filled = True,
        rounded = True
    )
plt.show()
```

## Unbekannten neuen Datensatz klassifizieren

```python
import numpy as np
messwerte = [[7.2, 3.0, 5.9, 2.0]]
messwerte = np.array(messwerte)
vorhersage = model.predict(messwerte)
print(vorhersage[0])
```

## Konfusionsmatrix erstellen

```python
from sklearn import metrics
import matplotlib.pyplot as plt
import numpy

vorhersagen = model.predict(X_test)
confusion_matrix = metrics.confusion_matrix(y_test, vorhersagen)

cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix, display_labels = ["Setosa ", "Versicolour", "Virginica"])
cm_display.plot()
plt.show()
```
